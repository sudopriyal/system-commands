# Argument - Parameter Extension

Variables in shell script

```python
%%bash
name="Priyal"
echo $name
```

```python
%%bash
name="Priyal"
echo ${name}Patel
```

```python
%%bash
bash greet.sh Priyal India
```

```python
%%bash
first="name"
name="Priyal"

echo ${!first}
```

Command Substitution

```python
%%bash
echo "This is the Present Working Directory: $(pwd)"
```

Arithmetic Expression

```python
%%bash
echo "Summation of 2 and 3 is $((2+3))"
```

```python
%%bash
touch file1.txt file2.txt file3.txt file4.pdf file5.pdf
files="*.txt"
echo $files
```

Brace Expansion

```python
%%bash
echo {1..5}
```

Echoing in Uppercase

```python
%%bash
name="Priyal"
echo "Hello, ${name^^}!"
```

Length of String

```python
%%bash
name="Priyal"
echo "Length of String: ${#name}"
```

Wildcards

`\*` (matches any number of characters, including none) <br>
`?` (matches a single character) <br>
`[...]` (matches any one of the enclosed characters) <br>

List all files ending in .txt

```python
%%bash
echo ls *.txt
```

List all files with one digit number

```python
%%bash
touch file10.txt
```

```python
%%bash
ls file?.txt
```

List all files within range 1 to 3

```python
%%bash
touch file4.txt file5.txt
```

```python
%%bash
ls file[1-3].txt
```
